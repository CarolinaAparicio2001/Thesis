```python
import torch
from torch.utils.data import DataLoader
from clinical_trial_dataset import ClinicalTrialDataset
from collate_fn import clinical_trial_collate
from Models.mpnn import MPNN
from Models.mmcto.mmcto import MMCTO
from Models.mmf.early_fusion import get_encoder_dict
from mmcto_interpretability import LightningModule as MMCTO_InterpWrapper
import pytorch_lightning as pl
import os
```

```python
dataset_name = "hint"  # or "ctod"
phase = "III"
model_checkpoint_path = "checkpoints/mmcto_hint_phaseIII.pt"  # Path to saved MMCTO model weights
predict_path = f"predictions/{dataset_name}_phase{phase}"  # Will store .pkl files here

# === Load dataset ===
data_dir = "data/clinical-trial-outcome-prediction/data_reduced" if dataset_name == "hint" else "data/labelling/labelling_reduced"
test_dataset = ClinicalTrialDataset(
    data_dir=data_dir,
    phase=phase,
    split="valid",
    dataset_name=dataset_name
)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False, collate_fn=clinical_trial_collate)

# === Load model ===
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
mpnn_model = MPNN(device=device)
encoders = get_encoder_dict(mpnn_model)
model = MMCTO(encoders=encoders).to(device)

# Load trained weights
if os.path.exists(model_checkpoint_path):
    model.load_state_dict(torch.load(model_checkpoint_path, map_location=device))
    print(f"Loaded checkpoint from {model_checkpoint_path}")
else:
    print(f"[!] Checkpoint not found: {model_checkpoint_path}")

# === Wrap model for interpretability ===
wrapped_model = MMCTO_InterpWrapper(model=model, predict_tasks=["Label"])
wrapped_model.predict_path = predict_path  # where .pkl will be saved

# === Run prediction ===
trainer = pl.Trainer(
    accelerator="gpu" if torch.cuda.is_available() else "cpu",
    devices=1,
    logger=False,
    enable_checkpointing=False,
)

trainer.predict(model=wrapped_model, dataloaders=test_loader)

print(f"Prediction + interpretability logging complete! Check: {predict_path}/hidden_state/")
```
