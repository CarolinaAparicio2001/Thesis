```python
import os
import json # Not actively used in the final script, but often useful for I/O
import yaml
import pandas as pd
from flask import Flask, request, jsonify
from flask_cors import CORS
# from sentence_transformers import SentenceTransformer # Replaced by HuggingFaceEmbeddings for better LangChain integration
from langchain.vectorstores import FAISS
from langchain.schema import Document
# from langchain.text_splitter import RecursiveCharacterTextSplitter # Not actively used, but can be useful for long documents
from langchain_huggingface import HuggingFaceEmbeddings
from langchain.prompts import PromptTemplate
from langchain.chains import RetrievalQA
from langchain_openai import ChatOpenAI
# import faiss # Direct faiss import not needed when using LangChain's FAISS wrapper
# import numpy as np # numpy might be needed for custom FAISS operations, but not for basic LangChain FAISS
# import pickle # For saving/loading docstore and index_to_docstore_id if doing manual FAISS management
```

```python
# --- Initial Configuration and API Key Loading ---
def load_api_key_from_yaml(path):
    try:
        with open(path, "r") as f:
            config = yaml.safe_load(f)
        key = config.get("api_key")
        if not key:
            raise ValueError("API key not found in YAML file.")
        return key
    except FileNotFoundError:
        print(f"WARNING: API key file not found at {path}. Please ensure the path is correct.")
        return None
    except Exception as e:
        print(f"Error loading API key from {path}: {e}")
        return None
```

```python
# --- Constants and Paths (ADJUST AS NEEDED) ---
PHASES = ["I", "II", "III"]
# ATTENTION: These paths are examples and will likely need to be adjusted for your local environment.
CSV_BASE_PATH = "C:/Users/Carol/Documents/Data Code and Deliverables/Data Group Part/Data/clinical-trial-outcome-prediction/Processed/hint/rag" 
API_KEY_PATH = r"C:/Users/Carol/Documents/Data Code and Deliverables/Data Group Part/openai_key.yaml"

FAISS_LANGCHAIN_STORE_PATH = "faiss_clinical_trials_store_en" # Folder to save the LangChain FAISS index and docstore

api_key = load_api_key_from_yaml(API_KEY_PATH)
if api_key:
    os.environ["OPENAI_API_KEY"] = api_key
    print("OpenAI API key loaded and set as environment variable.")
else:
    print("WARNING: OpenAI API key not loaded. The LLM will not function without it. Please set OPENAI_API_KEY manually or correct API_KEY_PATH.")

# --- Data Loading and Preprocessing Functions ---
def load_phase_data(phase, split="train"):
    csv_path = f"{CSV_BASE_PATH}/filtered_phase_{phase}_{split}.csv"
    try:
        csv_df = pd.read_csv(csv_path)
        # print(f"Columns in phase {phase}, {split}: {list(csv_df.columns)}") # Uncomment for debugging
        return csv_df
    except FileNotFoundError:
        print(f"WARNING: File not found: {csv_path}")
        return pd.DataFrame()
    except Exception as e:
        print(f"Error reading {csv_path}: {e}")
        return pd.DataFrame()

def clean_criteria(criteria):
    if isinstance(criteria, str):
        criteria = criteria.replace("\r\n", " ").replace("\n", " ").strip()
        criteria = ' '.join(criteria.split())
    return criteria

def load_all_data():
    all_dfs = []
    for phase_val in PHASES:
        for split_val in ["train", "test"]:
            df_phase_split = load_phase_data(phase_val, split_val)
            if not df_phase_split.empty:
                # Ensure all necessary columns exist, filling with 'N/A' if absent
                required_cols = ["NCT Number", "Enrollment", "criteria", "label", "smiles", "icdcodes", "fused_pred", "diseases", "drugs"]
                for col in required_cols:
                    if col not in df_phase_split.columns:
                        print(f"WARNING: Column '{col}' not found in phase {phase_val}, {split_val}. It will be filled with 'N/A'.")
                        df_phase_split[col] = "N/A"
                
                if "criteria" in df_phase_split.columns:
                     df_phase_split["criteria"] = df_phase_split["criteria"].apply(clean_criteria)
                else:
                    df_phase_split["criteria"] = "N/A" # If 'criteria' didn't exist and was added as N/A

                # Add the 'phase_meta' column for metadata, using the value from the phase_val variable
                df_phase_split["phase_meta"] = phase_val
                all_dfs.append(df_phase_split)
            # else:
                # print(f"Data not loaded for phase {phase_val} {split_val}. The file might be missing or empty.")
    
    if not all_dfs:
        print("CRITICAL ERROR: No data was loaded. Check CSV_BASE_PATH and the structure of your CSV files.")
        return pd.DataFrame()
    
    # Concatenate and then create combined_text to avoid SettingWithCopyWarning issues
    final_df = pd.concat(all_dfs, ignore_index=True)
    
    # Create 'combined_text' after concatenation and ensuring columns exist
    # Ensure all components of combined_text are strings
    final_df["combined_text"] = (
        final_df["NCT Number"].astype(str) + " " +
        final_df["Enrollment"].astype(str) + " " +
        final_df["criteria"].astype(str) + " " +
        final_df["label"].astype(str) + " " +
        final_df["smiles"].apply(lambda x: ' '.join(x) if isinstance(x, list) else str(x)) + " " +
        final_df["icdcodes"].apply(lambda x: ' '.join(x) if isinstance(x, list) else str(x)) + " " +
        final_df["fused_pred"].astype(str)
    )
    return final_df
```

```python
def create_langchain_documents(df):
    docs = []
    for index, row in df.iterrows():
        content = f"""
        Clinical trial NCT Number: {row.get('NCT Number', 'N/A')}.
        Phase: {row.get('phase_meta', 'N/A')}.
        Enrollment: {row.get('Enrollment', 'N/A')} participants.
        Diseases: {str(row.get('diseases', 'N/A'))}.
        Drugs: {str(row.get('drugs', 'N/A'))}.
        Eligibility Criteria: {row.get('criteria', 'N/A')}.
        ICD Codes: {str(row.get('icdcodes', 'N/A'))}.
        SMILES (Molecular Information): {str(row.get('smiles', 'N/A'))}.
        Original Model Prediction Score (fused_pred): {row.get('fused_pred', 'N/A')}.
        Actual Outcome (label - 1 for success, 0 for failure): {row.get('label', 'N/A')}.
        """
        # Add relevant metadata for filtering and context
        metadata = {
            "source": "csv_clinical_trials", 
            "nct_number": row.get("NCT Number", "N/A"), 
            "phase": row.get("phase_meta", "N/A"), 
            "label": row.get("label", "N/A"),
            "original_fused_pred": row.get("fused_pred", "N/A"),
            "row_index": index
        }
        docs.append(Document(page_content=content.strip(), metadata=metadata))
    return docs
```

```python
eh_model_name = 'sentence-transformers/all-MiniLM-L6-v2'
print(f"Using embedding model: {eh_model_name}")
embedding_function_hf = HuggingFaceEmbeddings(model_name=eh_model_name)

def save_lc_faiss_store(vector_store, path_base):
    try:
        vector_store.save_local(folder_path=path_base)
        print(f"LangChain FAISS vector store saved to {path_base}")
    except Exception as e:
        print(f"Error saving LangChain FAISS vector store: {e}")

def load_lc_faiss_store(path_base, embeddings_func):
    if os.path.exists(os.path.join(path_base, "index.faiss")) and os.path.exists(os.path.join(path_base, "index.pkl")):
        try:
            # For Langchain >0.2.0, allow_dangerous_deserialization might not be necessary or might have changed.
            # Check your Langchain version's documentation if you encounter issues here.
            vector_store = FAISS.load_local(folder_path=path_base, embeddings=embeddings_func, allow_dangerous_deserialization=True)
            print(f"LangChain FAISS vector store loaded from {path_base}")
            return vector_store
        except Exception as e:
            print(f"Error loading LangChain FAISS vector store from {path_base}: {e}")
            return None
    else:
        # print(f"WARNING: FAISS vector store files not found at {path_base}. The index will not be loaded.")
        return None
```

```python
df_main_data = load_all_data()
vector_store_global = None

if not df_main_data.empty:
    print(f"Total of {len(df_main_data)} records loaded and processed.")
    lc_documents = create_langchain_documents(df_main_data)
    print(f"Total of {len(lc_documents)} LangChain documents created.")

    if lc_documents:
        print(f"Attempting to load vector store from '{FAISS_LANGCHAIN_STORE_PATH}'...")
        vector_store_global = load_lc_faiss_store(FAISS_LANGCHAIN_STORE_PATH, embedding_function_hf)

        if vector_store_global is None:
            print(f"Vector store not found or error during loading. Creating a new one at '{FAISS_LANGCHAIN_STORE_PATH}'...")
            try:
                vector_store_global = FAISS.from_documents(lc_documents, embedding_function_hf)
                print("New FAISS vector store created.")
                save_lc_faiss_store(vector_store_global, FAISS_LANGCHAIN_STORE_PATH)
            except Exception as e:
                print(f"CRITICAL ERROR creating FAISS vector store: {e}")
        
        if vector_store_global:
             print(f"FAISS vector store ready for use. Number of vectors: {vector_store_global.index.ntotal}")
    else:
        print("No LangChain documents were created; cannot build the vector store.")
else:
    print("CRITICAL ERROR: Main DataFrame is empty. RAG cannot be initialized.")
```

```python
qa_chain_global = None
if vector_store_global and api_key: # Only configure if vector_store and API key are available
    try:
        llm = ChatOpenAI(temperature=0.1, model_name="gpt-3.5-turbo") # You can adjust this
        retriever = vector_store_global.as_retriever(search_kwargs={"k": 3}) # Retrieve top 3 most similar documents

        prompt_template_rag = """
        You are an expert assistant specializing in clinical trial analysis.
        Use the following retrieved CONTEXT from a clinical trials database to answer the QUESTION.
        If the context does not contain enough information to answer the question, explicitly state that you do not have sufficient information based on the provided documents.
        Do not attempt to invent an answer or use external knowledge.

        CONTEXT:
        {context}

        QUESTION: {question}

        Based ONLY on the provided context, answer as follows:
        1. PREDICTION OF TRIAL OUTCOME: (Indicate if the clinical trial described in the context is likely to SUCCEED or FAIL. Base this on qualitative data from the context. If the context mentions an 'Actual Outcome (label)' or 'Original Model Prediction (fused_pred)', you can use them as a reference, but form your own prediction based on the provided details.)
        2. DETAILED EXPLANATION OF THE PREDICTION: (Provide a detailed explanation for your prediction, citing the most important factors from the context, such as eligibility criteria, trial phase, diseases, drugs, sample size (enrollment), etc., that led you to your conclusion. If the 'Actual Outcome (label)' is available, compare your prediction with it.)
        3. KEY RISK AND SUCCESS FACTORS IDENTIFIED IN CONTEXT: (List the main risk factors for failure and the main factors that could contribute to the success of this specific trial, based strictly on the provided context.)

        ANALYTICAL RESPONSE:
        """
        RAG_PROMPT = PromptTemplate(
            template=prompt_template_rag, input_variables=["context", "question"]
        )

        qa_chain_global = RetrievalQA.from_chain_type(
            llm=llm,
            chain_type="stuff", 
            retriever=retriever,
            return_source_documents=True,
            chain_type_kwargs={"prompt": RAG_PROMPT}
        )
        print("RetrievalQA chain configured successfully.")
    except Exception as e:
        print(f"ERROR configuring RetrievalQA chain: {e}")
else:
    if not vector_store_global:
        print("WARNING: Vector store is not available. RetrievalQA chain will not be configured.")
    if not api_key:
        print("WARNING: OpenAI API key is not available. RetrievalQA chain will not be configured.")
```

```python
app = Flask(__name__)
CORS(app)

@app.route("/", methods=["GET"])
def root_route():
    return "<h3>Clinical Trials RAG API is running!</h3>Use the /v1/predict_trial endpoint to interact."

@app.route("/v1/predict_trial", methods=["POST"])
def handle_predict_trial():
    global qa_chain_global
    try:
        data = request.json
        user_query = data.get("query")

        if not user_query:
            return jsonify({"error": "'query' is required in the request body."}), 400

        if qa_chain_global is None:
            error_msg = "RAG chain (RetrievalQA) is not initialized."
            if not vector_store_global:
                error_msg += " The Vector Store was not loaded/created."
            if not api_key:
                error_msg += " The OpenAI API key is not configured."
            print(f"Error in /v1/predict_trial endpoint: {error_msg}")
            return jsonify({"error": error_msg}), 503 # Service Unavailable
        
        print(f"Received query for /v1/predict_trial: {user_query}")
        # Use .invoke() which returns a dictionary. The answer is in the 'result' key.
        response_dict = qa_chain_global.invoke({"query": user_query})
        
        final_answer = response_dict.get('result', 'Error: No answer found in the \'result\' key.')
        source_docs_content = []
        if response_dict.get('source_documents'):
            for doc in response_dict['source_documents']:
                source_docs_content.append({
                    "page_content": doc.page_content,
                    "metadata": doc.metadata
                })

        return jsonify({
            "query": user_query,
            "prediction_explanation": final_answer,
            "source_documents_retrieved": source_docs_content
        })
    
    except Exception as e:
        print(f"Error processing request at /v1/predict_trial: {e}")
        import traceback
        traceback.print_exc() # Print full traceback on the server for debugging
        return jsonify({"error": f"An internal server error occurred: {str(e)}"}), 500

if __name__ == "__main__":
    if not api_key:
        print("### CRITICAL ERROR: APPLICATION CANNOT START WITHOUT OPENAI API KEY ###")
        print("### Please configure API_KEY_PATH correctly or set the OPENAI_API_KEY environment variable ###")
    elif vector_store_global is None:
        print("### CRITICAL ERROR: APPLICATION CANNOT START WITHOUT A VALID VECTOR STORE ###")
        print("### Check logs for errors during FAISS index loading or creation. ###")
    elif qa_chain_global is None:
        print("### CRITICAL ERROR: APPLICATION CANNOT START BECAUSE THE RAG CHAIN WAS NOT CONFIGURED ###")
        print("### Check logs for errors in RetrievalQA configuration. ###")
    else:
        print("[INFO] Starting Flask server for Clinical Trials RAG API...")
        # use_reloader=False is important to prevent RAG initialization from running twice
        app.run(debug=False, port=5000, use_reloader=False, host='0.0.0.0') 
```
