```python
import sys
import os
import torch
import json
import numpy as np
import pytorch_lightning as pl
from sklearn.metrics import average_precision_score, roc_auc_score, f1_score
from torch.optim import Adam
from torch.utils.data import DataLoader
```

```python
# Add the path to your project folder where all your .py files are located
sys.path.append(os.path.abspath('../'))  # Adjust this path based on where your files are located
print(sys.path)

# Now import the converted Python modules (the ones from the .py files)
from mmcto import MMCTO
from mmf.early_fusion import EarlyFusion
from mmf.middle_fusion import MiddleFusion
from mmf.late_fusion import LateFusion
from torch.utils.data import Dataset
import pytorch_lightning.callbacks as pl_callbacks
```

```python
# Trainer settings (hardcoded as per previous config)
max_epochs = 10  # Max epochs as per config
batch_size = 32  # Batch size as per config
learning_rate = 1e-4  # Learning rate as per config
precision = 32  # Mixed precision setting
log_every_n_steps = 50  # Logging interval
```

```python
class CustomDataset(Dataset):
    def __init__(self, data_dir, phase, augment=False):
        # This is a placeholder for your data loading logic
        self.data_dir = data_dir
        self.phase = phase
        self.augment = augment
        # Add logic to load and process your data here
        self.data = []  # Replace with actual data
        self.labels = []  # Replace with actual labels

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        # Return a single item (feature, label)
        return {
            'input': self.data[idx],  # Replace with actual feature processing
            'label': self.labels[idx],  # Replace with actual label processing
        }

# Device setup (GPU if available)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```python
# Model checkpoint callback
checkpoint_callback = pl.callbacks.ModelCheckpoint(
    monitor="val/loss",
    filename="epoch:{epoch}-val_loss:{val/loss:.4g}",
    save_top_k=3,
    mode="min",
)

# Learning rate monitor callback
lr_monitor = pl.callbacks.LearningRateMonitor(logging_interval="step")
```

```python
# Set up the PyTorch Lightning trainer
trainer = pl.Trainer(
    max_epochs=max_epochs,
    devices="auto",
    precision=precision,
    callbacks=[checkpoint_callback, lr_monitor],
    log_every_n_steps=log_every_n_steps
)
```

```python
def get_loaders(data_path, phase="I", dataset_name="hint"):
    # Use the custom dataset class (you should adjust this to load your actual data)
    train_dataset = CustomDataset(data_dir=data_path, phase=phase, augment=True)
    val_dataset = CustomDataset(data_dir=data_path, phase=phase, augment=False)
    test_dataset = CustomDataset(data_dir=data_path, phase=phase, augment=False)
    
    # Create data loaders for train, validation, and test
    train_loader = DataLoader(
        train_dataset, 
        batch_size=batch_size, 
        shuffle=True
    )
    val_loader = DataLoader(
        val_dataset, 
        batch_size=batch_size, 
        shuffle=False
    )
    test_loader = DataLoader(
        test_dataset, 
        batch_size=batch_size, 
        shuffle=False
    )
    
    return train_loader, val_loader, test_loader
```

```python
def get_model(model_type="mmcto", fusion_dim=128):
    if model_type == "early":
        model = EarlyFusion(fusion_dim=fusion_dim)  # Use EarlyFusion directly
    elif model_type == "middle":
        model = MiddleFusion(fusion_dim=fusion_dim)  # Use MiddleFusion directly
    elif model_type == "late":
        model = LateFusion(fusion_dim=fusion_dim)  # Use LateFusion directly
    elif model_type == "mmcto":
        model = MMCTO()  # Use MMCTO directly
    else:
        raise ValueError(f"Unknown model_type: {model_type}")
    
    return model.to(device)
```

```python
def evaluate(model, loader):
    model.eval()
    preds, labels = [], []
    device = next(model.parameters()).device

    with torch.no_grad():
        for batch in loader:
            batch = {k: (v.to(device) if isinstance(v, torch.Tensor) else v) for k, v in batch.items()}
            output = model(batch)
            pred = output["metric_dict"]["preds"]
            
            preds.append(pred.cpu())
            labels.append(batch["label"].cpu())

    preds = torch.cat(preds)
    labels = torch.cat(labels)

    roc_auc = roc_auc_score(labels.numpy(), preds.numpy())
    pr_auc = average_precision_score(labels.numpy(), preds.numpy())
    f1 = f1_score(labels.numpy(), (preds.numpy() > 0.5).astype(int))

    return roc_auc, pr_auc, f1
```

```python
def train(model, train_loader, val_loader, epochs=5, lr=1e-4):
    optimizer = Adam(model.parameters(), lr=lr)
    device = next(model.parameters()).device
    best_val_score = -np.inf
    best_model = None

    for epoch in range(epochs):
        model.train()
        for batch in train_loader:
            batch = {k: (v.to(device) if isinstance(v, torch.Tensor) else v) for k, v in batch.items()}
            optimizer.zero_grad()
            output = model(batch)
            loss = output["loss_dict"]["loss"]

            loss.backward()
            optimizer.step()

        val_roc_auc, val_pr_auc, val_f1 = evaluate(model, val_loader)
        print(f"[Epoch {epoch}] Val ROC-AUC: {val_roc_auc:.4f}, PR-AUC: {val_pr_auc:.4f}, F1: {val_f1:.4f}")

        if val_pr_auc > best_val_score:
            best_val_score = val_pr_auc
            best_model = model.state_dict()

    return best_model
```

```python
def main():
    model_types = ["early", "middle", "late", "mmcto"]
    phases = ["I", "II", "III"]
    datasets = ["hint", "ctod"]

    results = {}

    # Paths to the datasets
    hint_data_path = r'C:/Users/Asus/Documents/Data and Code Deliverables/Data Group Part/project/datasets/hint'
    ctod_data_path = r'C:/Users/Asus/Documents/Data and Code Deliverables/Data Group Part/project/datasets/ctod'

    for dataset_name in datasets:
        if dataset_name == "hint":
            data_path = hint_data_path
        elif dataset_name == "ctod":
            data_path = ctod_data_path

        print(f"\n### Dataset: {dataset_name.upper()} ###")
        results = {}

        for model_type in model_types:
            results[model_type] = {}
            for phase in phases:
                print(f"\n==== Training {model_type.upper()} on Phase {phase} ====")
                train_loader, val_loader, test_loader = get_loaders(data_path, phase=phase, dataset_name=dataset_name)
                model = get_model(model_type=model_type)
                best_model = train(model, train_loader, val_loader, epochs=10)  # Removed test_loader here
                model.load_state_dict(best_model)

                # Evaluate on the validation set after training
                val_roc_auc, val_pr_auc, val_f1 = evaluate(model, val_loader)
                print(f"[Epoch {epoch}] Val ROC-AUC: {val_roc_auc:.4f}, PR-AUC: {val_pr_auc:.4f}, F1: {val_f1:.4f}")

                # Evaluate on the test set after training
                test_roc_auc, test_pr_auc, test_f1 = evaluate(model, test_loader)  # Evaluate on test set
                print(f"Test ROC-AUC: {test_roc_auc:.4f}, PR-AUC: {test_pr_auc:.4f}, F1: {test_f1:.4f}")
                
                results[model_type][phase] = (test_roc_auc, test_pr_auc, test_f1)

        print("\n=== Final Results ===")
        for model_type, phase_results in results.items():
            print(f"\nModel: {model_type.upper()}")
            for phase, (roc, pr, f1_) in phase_results.items():
                print(f"Phase {phase} -> ROC AUC: {roc:.4f}, PR AUC: {pr:.4f}, F1: {f1_:.4f}")

if __name__ == "__main__":
    main()
```
