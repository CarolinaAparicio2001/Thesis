```python
import os

def verify_data_prefix(data_prefix):
    checks = {
        "Data Path": os.path.isdir(data_prefix["data_path"]),
        "Table Path": os.path.isdir(data_prefix["table_path"]),
        "Summarization Path": os.path.isdir(data_prefix["summarization_path"]),
        "Drug Description Path": os.path.isfile(data_prefix["drug_description_path"]),
        "Criteria Path": os.path.isdir(data_prefix["criteria_path"]),
    }
    for name, exists in checks.items():
        print(f"{name}: {'✅ Exists' if exists else '❌ MISSING'}")


# Base path to your dataset root
base_path = r"C:/Users/Asus/Documents/Data and Code Deliverables/Data Group Part/Data/clinical-trial-outcome-prediction/Processed/ctod"

# Constructing full paths to each required directory or file
data_prefix = dict(
    data_path=base_path,
    table_path=os.path.join(base_path, "text_description"),
    summarization_path=os.path.join(base_path, "brief_summary"),
    drug_description_path=os.path.join(base_path, "drugbank", "druginfo_description.json"),
    criteria_path=os.path.join(base_path, "criteria"),
)

# Run the verification
verify_data_prefix(data_prefix)
```

```python
import os

# Adjust to your actual base path
base_path = r"C:/Users/Asus/Documents/Data and Code Deliverables/Data Group Part/Data/clinical-trial-outcome-prediction/Processed/hint"
splits = ["train", "valid", "test"]
phases = ["I", "II", "III"]

for phase in phases:
    for split in splits:
        tokenized_file = os.path.join(base_path, f"phase_{phase}_{split}_tokenized.pt")
        if os.path.exists(tokenized_file):
            os.remove(tokenized_file)
            print(f"Removed cache: {tokenized_file}")
        else:
            print(f"No cache found: {tokenized_file}")
```
