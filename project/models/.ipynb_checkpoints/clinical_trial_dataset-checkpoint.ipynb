```python
# === clinical_trial_dataset.py ===
import os
import torch
import pandas as pd
import numpy as np
from torch.utils.data import Dataset
from transformers import AutoTokenizer
from Models.data_utils import smiles2mpnnfeature

class ClinicalTrialDataset(Dataset):
    def __init__(self, data_dir, phase="I", split="train", dataset_name="hint", text_dir="text_description", summary_dir="brief_summary"):
        super().__init__()
        self.data_dir = data_dir
        self.dataset_name = dataset_name
        self.phase = phase
        self.split = split

        self.df = pd.read_csv(os.path.join(data_dir, f"phase_{phase}_{split}.csv"))

        if dataset_name == "hint":
            base_dir = "data/clinical-trial-outcome-prediction/data"
        elif dataset_name == "ctod":
            base_dir = "data/labelling"
        else:
            raise ValueError("dataset_name must be hint or ctod")

        criteria_path = os.path.join(base_dir, "criteria", f"phase_{phase}_{split}.npy")
        self.criteria_features = torch.tensor(np.load(criteria_path), dtype=torch.float32)

        self.description_dir = os.path.join(data_dir, text_dir)
        self.summary_dir = os.path.join(data_dir, summary_dir)

        self.tokenizer = AutoTokenizer.from_pretrained("medicalai/ClinicalBERT")

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        label = torch.tensor(row["Label"], dtype=torch.float32)
        criteria = self.criteria_features[idx]
        smiles = row["SMILES"]
        smiles_feat = smiles2mpnnfeature(smiles)

        desc_path = os.path.join(self.description_dir, f"phase_{self.phase}_{self.split}.csv")
        desc_text = pd.read_csv(desc_path).iloc[idx]["description"]
        desc_inputs = self.tokenizer(desc_text, return_tensors="pt", truncation=True, padding="max_length", max_length=128)

        summ_path = os.path.join(self.summary_dir, f"phase_{self.phase}_{self.split}.csv")
        summ_text = pd.read_csv(summ_path).iloc[idx]["summary"]
        summ_inputs = self.tokenizer(summ_text, return_tensors="pt", truncation=True, padding="max_length", max_length=128)

        return {
            "criteria": criteria,
            "smiless": smiles_feat,
            "description": desc_inputs,
            "summarization": summ_inputs,
            "label": label,
        }
```
