```python
import os
import json
import yaml
import torch
import pandas as pd
import pickle
from langchain.vectorstores import FAISS
from langchain_openai import OpenAIEmbeddings
from langchain.docstore.document import Document
from langchain.chat_models import ChatOpenAI
from langchain.schema import SystemMessage, HumanMessage
from langchain.text_splitter import RecursiveCharacterTextSplitter
```

```python
def load_api_key_from_yaml(path=r"C:/Users/Carol/Documents/Data Code and Deliverables/Data Group Part/openai_key.yaml"):
    with open(path, "r") as f:
        config = yaml.safe_load(f)
    key = config.get("api_key")
    if not key:
        raise ValueError("API key not found in YAML file.")
    return key
```

```python
def load_text_descriptions(base_dir, dataset_name):
    path = os.path.join(base_dir, dataset_name, "text_description")
    if not os.path.exists(path):
        print(f"\u274c Not found: {path}")
        return []

    documents = []
    for filename in os.listdir(path):
        if filename.endswith(".json"):
            phase_name = filename.replace(".json", "")
            file_path = os.path.join(path, filename)
            with open(file_path, "r", encoding="utf-8") as f:
                try:
                    entries = json.load(f)
                except Exception as e:
                    print(f"[!] Failed to load {filename}: {e}")
                    continue
                for i, entry in enumerate(entries):
                    if isinstance(entry, str) and len(entry.strip()) > 0:
                        documents.append(Document(
                            page_content=entry,
                            metadata={
                                "dataset": dataset_name,
                                "phase": phase_name,
                                "index": i,
                                "source": "json"
                            }
                        ))
    print(f"\u2705 Loaded {len(documents)} documents from {dataset_name} text_description")
    return documents
```

```python
def load_csv_as_documents(base_dir, dataset_name):
    path = os.path.join(base_dir, dataset_name)
    documents = []
    for filename in os.listdir(path):
        if filename.endswith(".csv") and filename.startswith("phase"):
            file_path = os.path.join(path, filename)
            try:
                df = pd.read_csv(file_path)
                for i, row in df.iterrows():
                    summary = []
                    for col in ["phase", "diseases", "drugs", "criteria", "enrollment"]:
                        if col in row and pd.notna(row[col]):
                            summary.append(f"{col.capitalize()}: {row[col]}")
                    if summary:
                        documents.append(Document(
                            page_content="; ".join(summary),
                            metadata={
                                "dataset": dataset_name,
                                "phase_file": filename,
                                "index": i,
                                "source": "csv"
                            }
                        ))
            except Exception as e:
                print(f"[!] Failed to process {filename}: {e}")
    print(f"\u2705 Loaded {len(documents)} documents from {dataset_name} CSVs")
    return documents
```

```python
def save_chunks(docs, path):
    with open(path, "wb") as f:
        pickle.dump(docs, f)

def load_chunks(path):
    with open(path, "rb") as f:
        return pickle.load(f)
        
def build_faiss_index(sample_limit=100):
    base_path = r"C:/Users/Carol/Documents/Data Code and Deliverables/Data Group Part/Data/clinical-trial-outcome-prediction/Processed"
    datasets = ["hint"]
    api_key = load_api_key_from_yaml()

    all_documents = []
    for ds in datasets:
        docs_json = load_text_descriptions(base_path, ds)[:sample_limit//2]  # Limit per dataset
        docs_csv = load_csv_as_documents(base_path, ds)[:sample_limit//2]
        all_documents.extend(docs_json + docs_csv)

    print(f"\n Using {len(all_documents)} documents for FAISS index (sampled)")

    chunks_path = os.path.join(base_path, "cached_chunks_dev.pkl")
    if os.path.exists(chunks_path):
        print("⚡ Loading cached chunks (dev)...")
        split_docs = load_chunks(chunks_path)
    else:
        splitter = RecursiveCharacterTextSplitter(
            chunk_size=1000,
            chunk_overlap=100,
            separators=["\n\n", "\n", ".", ";", " ", ""]
        )
        split_docs = splitter.split_documents(all_documents)
        save_chunks(split_docs, chunks_path)
        print(f"📄 Chunked and cached {len(split_docs)} documents (dev subset)")

    index_path = os.path.join(base_path, "dev_faiss_index")
    embedding = OpenAIEmbeddings(model="text-embedding-3-small", openai_api_key=api_key)
    db = FAISS.from_documents(split_docs, embedding)
    db.save_local(index_path)
    print(f"FAISS index (dev) saved to: {index_path}")


#Predict binary success/fail
def predict_outcome(metric_dict):
    return 1 if metric_dict.get("preds", 0.0) > 0.5 else 0
```

```python
def retrieve_and_explain(query, metric_dict):
    base_path = r"C:/Users/Carol/Documents/Data Code and Deliverables/Data Group Part/Data/clinical-trial-outcome-prediction/Processed"
    index_path = os.path.join(base_path, "combined_faiss_index")
    api_key = load_api_key_from_yaml()

    embedding = OpenAIEmbeddings(model="text-embedding-3-small", openai_api_key=api_key)
    db = FAISS.load_local(index_path, embedding, allow_dangerous_deserialization=True)

    similar_docs = db.similarity_search(query, k=3)
    context = "\n\n".join([doc.page_content for doc in similar_docs])

    prediction = predict_outcome(metric_dict)
    result = "succeed" if prediction == 1 else "fail"

    prompt = f"""
The model predicts this trial is likely to {result}.

Query: {query}

Here are similar past trials from CTOD and HINT datasets:
{context}

Based on this, explain why the current trial is likely to {result}. Consider factors like enrollment, eligibility criteria, drug effectiveness, or disease complexity.
"""

    llm = ChatOpenAI(model="gpt-3.5-turbo", temperature=0, openai_api_key=api_key)
    messages = [
        SystemMessage(content="You are a biomedical expert helping explain why a clinical trial may succeed or fail."),
        HumanMessage(content=prompt)
    ]
    response = llm(messages)
    print("\n💬 Explanation:\n")
    print(response.content)

# ✅ Entry point
if __name__ == "__main__":
    build_faiss_index(sample_limit = 50)
    test_query = "Phase III study for rare cancer with low enrollment and strict eligibility criteria."
    mock_metric_dict = {"preds": torch.tensor(0.3)}
    retrieve_and_explain(test_query, mock_metric_dict)
```

```python
import os
import json
import torch
import pandas as pd
import numpy as np
import faiss
from sentence_transformers import SentenceTransformer
import openai
from rouge_score import rouge_scorer
from datasets.ctod.ctod_enrollment import CTODataset
from mmcto.mmcto import MMCTO
from mmcto.layers.sparse_moe import SparseMOELayer, FeedForwardLayer
from torch.utils.data import DataLoader
import yaml
```

```python
# === Set device ===
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# === Load OpenAI API key ===
def load_api_key_from_yaml(path=r"C:/Users/Carol/Documents/Data Code and Deliverables/Data Group Part/openai_key.yaml"):
    with open(path, "r") as f:
        config = yaml.safe_load(f)
    key = config.get("api_key")1
    if not key:
        raise ValueError("API key not found in YAML file.")
    return key


openai.api_key = load_api_key_from_yaml()
```

```python
PHASES = ["I", "II", "III"]
MODEL_PATH = "all-MiniLM-L6-v2"
PHASE_MODEL_PATHS = {
    "I": "C:/Users/Carol/Documents/Data Code and Deliverables/Data Group Part/project/models/mmcto_phaseI.pth",
    "II": "C:/Users/Carol/Documents/Data Code and Deliverables/Data Group Part/project/models/mmcto_phaseII.pth",
    "III": "C:/Users/Carol/Documents/Data Code and Deliverables/Data Group Part/project/models/mmcto_phaseIII.pth",
}
CSV_BASE_PATH = "C:/Users/Carol/Documents/Data Code and Deliverables/Data Group Part/Data/clinical-trial-outcome-prediction/Processed/hint/rag"
JSON_BASE_PATH = f"{CSV_BASE_PATH}/text_description"

# --- Sentence Embedding Model ---
embed_model = SentenceTransformer(MODEL_PATH)

# --- Prediction Model Config ---
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
MODEL_DIM = 768
```

```python
def combine_text(row):
    fields = [
        str(row.get("NCT Number", "")),
        str(row.get("Enrollment", "")),
        str(row.get("Eligibility Criteria", "")),
        str(row.get("summary", "")),
        " ".join(row.get("SMILES", []) if isinstance(row.get("SMILES"), list) else [str(row.get("SMILES", ""))]),
        " ".join(row.get("icd_code", []) if isinstance(row.get("icd_code"), list) else [str(row.get("icd_code", ""))]),
    ]
    return " ".join(fields)

def chunk_text(text, max_words=100):
    words = text.split()
    return [" ".join(words[i:i+max_words]) for i in range(0, len(words), max_words)]

def load_phase_data_with_predictions(phase):
    csv_file = os.path.join(CSV_BASE_PATH, f"filtered_phase_{phase}_test.csv")
    json_file = os.path.join(JSON_BASE_PATH, f"phase_{phase}_test.json")
    df = pd.read_csv(csv_file)

    if os.path.exists(json_file):
        with open(json_file, 'r') as f:
            summaries = json.load(f)
        df["summary"] = df.index.map(lambda i: summaries.get(str(i), ""))
    else:
        df["summary"] = ""

    dataset = CTODataset(
        ann_file_name=f"phase_{phase}_test",
        data_prefix=dict(
            data_path=CSV_BASE_PATH,
            table_path=f"{CSV_BASE_PATH}/text_description",
            summarization_path=f"{CSV_BASE_PATH}/brief_summary",
            drug_description_path=f"{CSV_BASE_PATH}/drugbank/druginfo_description.json",
            criteria_path=f"{CSV_BASE_PATH}/criteria",
        )
    )
    dataloader = DataLoader(dataset, batch_size=16, collate_fn=dataset.collate_fn)

    encoders = torch.nn.ModuleDict({k: torch.nn.TransformerEncoder(
        torch.nn.TransformerEncoderLayer(d_model=MODEL_DIM, nhead=8, dim_feedforward=2048, batch_first=True), num_layers=2
    ) for k in dataset.input_parts})

    smoe_encoder = SparseMOELayer(expert_cfg=FeedForwardLayer, num_experts=4, input_dim=MODEL_DIM, topk=2)
    model = MMCTO(encoders=encoders, smoe_encoder=smoe_encoder, model_dim=MODEL_DIM).to(DEVICE)
    model.load_state_dict(torch.load(PHASE_MODEL_PATHS[phase], map_location=DEVICE))
    model.eval()

    preds = []
    with torch.no_grad():
        for batch in dataloader:
            batch = {k: v.to(DEVICE) if torch.is_tensor(v) else v for k, v in batch.items()}
            out = model(batch)
            preds.extend(out["metric_dict"]["preds"].cpu().numpy())

    df["prediction"] = preds
    return df[df["prediction"] >= 0.8]
```

```python
# --- Build corpus from all phases ---
all_chunks, all_metadata = [], []
for phase in PHASES:
    df = load_phase_data_with_predictions(phase)
    df['combined_text'] = df.apply(combine_text, axis=1)
    for idx, row in df.iterrows():
        chunked = chunk_text(row['combined_text'])
        for i, ch in enumerate(chunked):
            all_chunks.append(ch)
            all_metadata.append({"nct": row['NCT Number'], "chunk_index": i, "phase": phase})

# --- Embedding & Indexing ---
embeddings = embed_model.encode(all_chunks, convert_to_numpy=True, normalize_embeddings=True)
index = faiss.IndexFlatIP(embeddings.shape[1])
index.add(np.ascontiguousarray(embeddings, dtype=np.float32))

# --- Querying ---
query = "I’m designing a trial for Crohn’s disease. Would enrolling 800 patients likely result in a prediction over 0.8 based on past trials?"
query_embedding = embed_model.encode([query], convert_to_numpy=True, normalize_embeddings=True)
dists, idxs = index.search(query_embedding, 5)
retrieved_chunks = [all_chunks[i] for i in idxs[0]]
retrieved_metadata = [all_metadata[i] for i in idxs[0]]
```

```python
# --- GPT-4o generation ---
def generate_with_gpt4o(retrieved_chunks, query):
    context = "\n\n".join(retrieved_chunks)
    prompt = f"Query: {query}\n\nRelevant chunks:\n{context}\n\nAnswer based on the trials above. Would 800 patients likely reach a prediction score of 0.8+? Justify your response."
    try:
        res = openai.ChatCompletion.create(
            model="gpt-4",
            messages=[
                {"role": "system", "content": "You are a clinical trial analysis assistant."},
                {"role": "user", "content": prompt},
            ],
            max_tokens=500,
            temperature=0.7,
        )
        return res['choices'][0]['message']['content'].strip()
    except Exception as e:
        print(f"OpenAI Error: {e}")
        return None

response = generate_with_gpt4o(retrieved_chunks, query)
print("\nGenerated Response:\n", response)
```
