```python
def load_api_key_from_yaml(path):
    with open(path, "r") as f:
        config = yaml.safe_load(f)
    key = config.get("api_key")
    if not key:
        raise ValueError("API key not found in YAML file.")
    return key
```

```python
PHASES = ["I", "II", "III"]
CSV_BASE_PATH = "C:/Users/Carol/Documents/Data Code and Deliverables/Data Group Part/Data/clinical-trial-outcome-prediction/Processed/hint/rag"
API_KEY_PATH = r"C:/Users/Carol/Documents/Data Code and Deliverables/Data Group Part/openai_key.yaml"
api_key = load_api_key_from_yaml(API_KEY_PATH)

FAISS_PATH = "clinical_trials_index"

# --- Load and Filter CSV ---
def load_phase_data(phase, split="train"):
    csv_path = f"{CSV_BASE_PATH}/filtered_phase_{phase}_{split}.csv"
    csv_df = pd.read_csv(csv_path)
    
    # Print column names to check if 'criteria' exists
    print(f"Columns in phase {phase}, {split}:", csv_df.columns)
    
    return csv_df

def clean_criteria(criteria):
    # Replace newline characters and excess whitespace
    if isinstance(criteria, str):
        # Remove newlines and extra spaces
        criteria = criteria.replace("\r\n", " ").replace("\n", " ").strip()
        criteria = ' '.join(criteria.split())  # Normalize spaces
    return criteria

def load_all_data():
    all_dfs = []
    for phase in ["I", "II", "III"]:
        for split in ["train", "test"]:
            try:
                df = load_phase_data(phase, split)
                # Ensure 'criteria' exists before applying cleaning
                if 'criteria' not in df.columns:
                    print(f"Column 'criteria' not found in phase {phase}, {split}. Available columns: {df.columns}")
                    continue
                
                # Clean the 'criteria' field before combining
                df['criteria'] = df['criteria'].apply(clean_criteria)
                
                # Ensure the 'combined_text' is created for embeddings
                df['combined_text'] = (
                    df['NCT Number'].astype(str) + " " +
                    df['Enrollment'].astype(str) + " " +
                    df['criteria'] + " " +  # Cleaned criteria
                    df['label'].astype(str) + " " +
                    df['smiles'].apply(lambda x: " ".join(x) if isinstance(x, list) else x) + " " +
                    df['icdcodes'].apply(lambda x: " ".join(x) if isinstance(x, list) else x) + " " +
                    df['fused_pred'].astype(str)
                )
                all_dfs.append(df)
            except FileNotFoundError:
                print(f"Missing phase {phase} {split}")
    
    return pd.concat(all_dfs, ignore_index=True)
```

```python
def create_documents(df):
    docs = []
    for _, row in df.iterrows():
        # Create CSV content (keep the structure intact, but as text)
        csv_content = f"""
        Clinical trial NCT Number is {row.get('NCT Number', 'N/A')}.
        It is a phase {row.get('phase', 'N/A')} trial with an enrollment of {row.get('enrollment', 'N/A')} participants.
        The trial is focused on the following diseases: {row.get('diseases', 'N/A')} and involves the drugs: {row.get('drugs', 'N/A')}.
        Eligibility criteria (from CSV) include: {row.get('criteria', 'N/A')}.
        The associated ICD codes are: {row.get('icdcodes', 'N/A')} and molecular information includes SMILES: {row.get('smiles', 'N/A')}.
        The model assigned this trial a prediction score of {row.get('fused_pred', 'N/A')}.
        """
        docs.append(Document(page_content=csv_content.strip(), metadata={"source": "csv"}))
    return docs
```

```python
def create_embeddings(df):
    model = SentenceTransformer('all-MiniLM-L6-v2')
    # Ensure no NaN or non-string values remain
    df['combined_text'] = df['combined_text'].apply(str)
    embeddings = model.encode(df['combined_text'].tolist(), convert_to_numpy=True)
    return embeddings
```

```python
def build_faiss_index(embeddings, docs):
    dimension = embeddings.shape[1]
    index = faiss.IndexFlatL2(dimension)
    embeddings = np.ascontiguousarray(embeddings, dtype=np.float32)
    index.add(embeddings)
    docstore = {i: docs[i].page_content for i in range(len(docs))}
    index_to_docstore_id = {i: i for i in range(len(docs))}
    return index, docstore, index_to_docstore_id

def save_faiss_index(index, path):
    faiss.write_index(index, path)

def load_faiss_index(path):
    try:
        return faiss.read_index(path)
    except Exception as e:
        print(f"Error loading FAISS index: {e}")
        return None
```

```python
def embed_question(question, index, df):
    # Generate the embedding for the question using the same model
    model = SentenceTransformer('all-MiniLM-L6-v2')
    query_embedding = model.encode([question], convert_to_numpy=True)
    
    # Perform search in FAISS to retrieve the top k most similar documents
    k = 5  # Number of similar documents to retrieve
    distances, indices = index.search(query_embedding, k)
    
    # Retrieve the corresponding documents from your DataFrame based on the indices
    retrieved_documents = []
    for i in indices[0]:
        doc = df.iloc[i]['combined_text']  # Access combined_text here
        retrieved_documents.append(doc)
    
    # Return the combined context of retrieved documents
    return "\n\n".join(retrieved_documents)
```

```python
# Step 1: Load the data
df = load_all_data()

# Step 2: Create documents
docs = create_documents(df)

# Step 3: Generate embeddings
embeddings = create_embeddings(df)

# Step 4: Build the FAISS index
index, docstore, index_to_docstore_id = build_faiss_index(embeddings, docs)

# Step 5: Save the FAISS index to disk
save_faiss_index(index, FAISS_PATH)

# Step 6: Load the FAISS index to verify
loaded_index = load_faiss_index(FAISS_PATH)
if loaded_index is not None:
    print("FAISS Index successfully loaded!")
```

```python
# Check the number of vectors in the index and the dimension of the embeddings
print("Number of vectors in FAISS index:", loaded_index.ntotal)
print("Dimension of each vector:", loaded_index.d)

# Retrieve and display the first few vectors in the index
# We will retrieve the first 5 vectors for illustration
k = 5
D, I = loaded_index.search(np.zeros((k, loaded_index.d), dtype=np.float32), k)

# Display the retrieved vectors and their corresponding distances
for idx, (distance, index) in enumerate(zip(D[0], I[0])):
    print(f"Vector {idx+1}:")
    print(f"Index: {index}")
    print(f"Distance: {distance}")
    print("------")
```

```python
# Retrieve the actual content (documents) corresponding to the first 5 vectors
for idx, (distance, index) in enumerate(zip(D[0], I[0])):
    doc_id = index_to_docstore_id.get(index, None)
    if doc_id is not None:
        print(f"Vector {idx+1} - Document:")
        print(f"Index: {index}")
        print(f"Distance: {distance}")
        print(f"Document: {docstore[doc_id]}")
        print("------")
```

```python
template = """
Query: {question}
Relevant documents: {context}

Provide the most relevant answer based on the information in the retrieved documents and the conversation history.
"""
prompt = PromptTemplate(input_variables=["question", "context"], template=template)

# Initialize ChatOpenAI (LLM initialization)
chat_model = ChatOpenAI(
    model="gpt-4",  # GPT-4 model
    openai_api_key=api_key,
    temperature=0.7  # Control creativity of responses
)

# Chat function that integrates OpenAI API
def chat(user_message, context):
    # Format the prompt using PromptTemplate
    formatted_prompt = prompt.format(question=user_message, context=context)
    
    # Create the list of messages expected by OpenAI
    messages = [
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": user_message},
        {"role": "assistant", "content": context},
    ]
    
    # Make a request to OpenAI's API
    response = chat_model.generate(messages)

    # Return the response from OpenAI
    return response[0].text.strip()
```
