```python
import os
import json
import yaml
import pandas as pd
from flask import Flask, request, jsonify
from flask_cors import CORS
from sentence_transformers import SentenceTransformer
from langchain.vectorstores import FAISS
from langchain.schema import Document
# from langchain.text_splitter import RecursiveCharacterTextSplitter # Não usado ativamente, mas pode ser útil
from langchain_huggingface import HuggingFaceEmbeddings
from langchain.prompts import PromptTemplate
from langchain.chains import RetrievalQA
from langchain_openai import ChatOpenAI
import faiss
import numpy as np
import pickle
```

```python
def load_api_key_from_yaml(path):
    try:
        with open(path, "r") as f:
            config = yaml.safe_load(f)
        key = config.get("api_key")
        if not key:
            raise ValueError("API key not found in YAML file.")
        return key
    except FileNotFoundError:
        print(f"Warning: File not found in {path}. Check if the path is correct.")
        return None
    except Exception as e:
        print(f"Error loading api key from {path}: {e}")
        return None
```

```python
PHASES = ["I", "II", "III"]
CSV_BASE_PATH = "C:/Users/Carol/Documents/Data Code and Deliverables/Data Group Part/Data/clinical-trial-outcome-prediction/Processed/hint/rag" 
API_KEY_PATH = r"C:/Users/Carol/Documents/Data Code and Deliverables/Data Group Part/openai_key.yaml"

FAISS_LANGCHAIN_STORE_PATH = "faiss_clinical_trials_store" # store FAISS index and langchain docstore
api_key = load_api_key_from_yaml(API_KEY_PATH)

# --- Funções de Carregamento e Pré-processamento de Dados ---
def load_phase_data(phase, split="train"):
    csv_path = f"{CSV_BASE_PATH}/filtered_phase_{phase}_{split}.csv"
    try:
        csv_df = pd.read_csv(csv_path)
        return csv_df
    except FileNotFoundError:
        print(f"Warning: File not found: {csv_path}")
        return pd.DataFrame()
    except Exception as e:
        print(f"Error reading {csv_path}: {e}")
        return pd.DataFrame()

def clean_criteria(criteria):
    if isinstance(criteria, str):
        criteria = criteria.replace("\r\n", " ").replace("\n", " ").strip()
        criteria = ' '.join(criteria.split())
    return criteria

def load_all_data():
    all_dfs = []
    for phase_val in PHASES:
        for split_val in ["train", "test"]:
            df_phase_split = load_phase_data(phase_val, split_val)
            if not df_phase_split.empty:
                # Guarantee all columns exist and replace nan values with N/A
                required_cols = ['NCT Number', 'Enrollment', 'criteria', 'label', 'smiles', 'icdcodes', 'fused_pred', 'diseases', 'drugs']
                for col in required_cols:
                    if col not in df_phase_split.columns:
                        print(f"Warning: Column '{col}' not found phase {phase_val}, {split_val}. will be filled with 'N/A'.")
                        df_phase_split[col] = 'N/A'
                
                if 'criteria' in df_phase_split.columns:
                     df_phase_split['criteria'] = df_phase_split['criteria'].apply(clean_criteria)
                else:
                    df_phase_split['criteria'] = 'N/A' # Se 'criteria' não existia e foi adicionada como N/A

                # Adicionar a coluna 'phase_meta' para metadados, usando o valor da variável phase_val
                df_phase_split['phase_meta'] = phase_val
                all_dfs.append(df_phase_split)
            # else:
                # print(f"Dados não carregados para fase {phase_val} {split_val}. O ficheiro pode estar em falta ou vazio.")
    
    if not all_dfs:
        print("ERRO CRÍTICO: Nenhum dado foi carregado. Verifique CSV_BASE_PATH e a estrutura dos seus ficheiros CSV.")
        return pd.DataFrame()
    
    # Concatenar e depois criar combined_text para evitar problemas com SettingWithCopyWarning
    final_df = pd.concat(all_dfs, ignore_index=True)
    
    # Criar 'combined_text' após a concatenação e garantia de colunas
    # Assegurar que todos os componentes de combined_text são strings
    final_df['combined_text'] = (
        final_df['NCT Number'].astype(str) + " " +
        final_df['Enrollment'].astype(str) + " " +
        final_df['criteria'].astype(str) + " " +
        final_df['label'].astype(str) + " " +
        final_df['smiles'].apply(lambda x: ' '.join(x) if isinstance(x, list) else str(x)) + " " +
        final_df['icdcodes'].apply(lambda x: ' '.join(x) if isinstance(x, list) else str(x)) + " " +
        final_df['fused_pred'].astype(str)
    )
    return final_df
```

```python
def create_langchain_documents(df):
    docs = []
    for index, row in df.iterrows():
        content = f"""
        Clinical trial NCT Number: {row.get('NCT Number', 'N/A')}.
        Phase: {row.get('phase_meta', 'N/A')}.
        Enrollment: {row.get('Enrollment', 'N/A')} participants.
        Diseases: {str(row.get('diseases', 'N/A'))}.
        Drugs: {str(row.get('drugs', 'N/A'))}.
        Eligibility Criteria: {row.get('criteria', 'N/A')}.
        ICD Codes: {str(row.get('icdcodes', 'N/A'))}.
        SMILES (Molecular Information): {str(row.get('smiles', 'N/A'))}.
        Original Model Prediction Score (fused_pred): {row.get('fused_pred', 'N/A')}.
        Actual Outcome (label - 1 for success, 0 for failure): {row.get('label', 'N/A')}.
        """
        # Adicionar metadados relevantes para filtragem e contexto
        metadata = {
            "source": "csv_clinical_trials", 
            "nct_number": row.get('NCT Number', 'N/A'), 
            "phase": row.get('phase_meta', 'N/A'), 
            "label": row.get('label', 'N/A'),
            "original_fused_pred": row.get('fused_pred', 'N/A'),
            "row_index": index
        }
        docs.append(Document(page_content=content.strip(), metadata=metadata))
    return docs
```

```python
eh_model_name = 'sentence-transformers/all-MiniLM-L6-v2'
print(f"A usar o modelo de embedding: {eh_model_name}")
embedding_function_hf = HuggingFaceEmbeddings(model_name=eh_model_name)

def save_lc_faiss_store(vector_store, path_base):
    try:
        vector_store.save_local(folder_path=path_base)
        print(f"Vector store LangChain FAISS guardado em {path_base}")
    except Exception as e:
        print(f"Erro ao guardar o vector store LangChain FAISS: {e}")

def load_lc_faiss_store(path_base, embeddings_func):
    if os.path.exists(os.path.join(path_base, "index.faiss")) and os.path.exists(os.path.join(path_base, "index.pkl")):
        try:
            # Para Langchain >0.2.0, allow_dangerous_deserialization pode não ser mais necessário ou pode ter mudado
            # Verifique a documentação da sua versão Langchain se encontrar problemas aqui.
            vector_store = FAISS.load_local(folder_path=path_base, embeddings=embeddings_func, allow_dangerous_deserialization=True)
            print(f"Vector store LangChain FAISS carregado de {path_base}")
            return vector_store
        except Exception as e:
            print(f"Erro ao carregar o vector store LangChain FAISS de {path_base}: {e}")
            return None
    else:
        # print(f"AVISO: Ficheiros do vector store FAISS não encontrados em {path_base}. O índice não será carregado.")
        return None
```

```python
df_main_data = load_all_data()
vector_store_global = None

if not df_main_data.empty:
    print(f"Total de {len(df_main_data)} registos carregados e processados.")
    lc_documents = create_langchain_documents(df_main_data)
    print(f"Total de {len(lc_documents)} documentos LangChain criados.")

    if lc_documents:
        print(f"A tentar carregar o vector store de '{FAISS_LANGCHAIN_STORE_PATH}'...")
        vector_store_global = load_lc_faiss_store(FAISS_LANGCHAIN_STORE_PATH, embedding_function_hf)

        if vector_store_global is None:
            print(f"Vector store não encontrado ou erro ao carregar. A criar um novo em '{FAISS_LANGCHAIN_STORE_PATH}'...")
            try:
                vector_store_global = FAISS.from_documents(lc_documents, embedding_function_hf)
                print("Novo vector store FAISS criado.")
                save_lc_faiss_store(vector_store_global, FAISS_LANGCHAIN_STORE_PATH)
            except Exception as e:
                print(f"ERRO CRÍTICO ao criar o vector store FAISS: {e}")
        
        if vector_store_global:
             print(f"Vector store FAISS pronto para uso. Número de vetores: {vector_store_global.index.ntotal}")
    else:
        print("Nenhum documento LangChain foi criado, não é possível construir o vector store.")
else:
    print("ERRO CRÍTICO: DataFrame principal está vazio. O RAG não pode ser inicializado.")
```

```python
qa_chain_global = None
if vector_store_global and api_key: # Só configurar se o vector_store e a API key estiverem disponíveis
    try:
        llm = ChatOpenAI(temperature=0.4, model_name="gpt-3.5-turbo", openai_api_key=api_key) # Pode ajustar
        retriever = vector_store_global.as_retriever(search_kwargs={"k": 3}) # Recuperar os 3 documentos mais similares

        prompt_template_rag = """
        Você é um assistente especialista em análise de ensaios clínicos.
        Use o seguinte CONTEXTO recuperado de uma base de dados de ensaios clínicos para responder à PERGUNTA.
        Se o contexto não contiver informação suficiente para responder à pergunta, diga explicitamente que não tem informação suficiente com base nos documentos fornecidos.
        Não tente inventar uma resposta ou usar conhecimento externo.

        CONTEXTO:
        {context}

        PERGUNTA: {question}

        Com base APENAS no contexto fornecido, responda da seguinte forma:
        1. PREVISÃO DO RESULTADO DO ENSAIO: (Indique se o ensaio clínico descrito no contexto provavelmente terá SUCESSO ou FALHARÁ. Baseie-se nos dados qualitativos do contexto. Se o contexto mencionar um 'Actual Outcome (label)' ou 'Original Model Prediction (fused_pred)', pode usá-los como referência, mas forme a sua própria previsão com base nos detalhes fornecidos.)
        2. EXPLICAÇÃO DETALHADA DA PREVISÃO: (Forneça uma explicação detalhada para a sua previsão, citando os fatores mais importantes do contexto, como critérios de elegibilidade, fase do ensaio, doenças, medicamentos, tamanho da amostra (enrollment), etc., que o levaram à sua conclusão. Se o 'Actual Outcome (label)' estiver disponível, compare a sua previsão com ele.)
        3. PRINCIPAIS FATORES DE RISCO E SUCESSO IDENTIFICADOS NO CONTEXTO: (Liste os principais fatores de risco para falha e os principais fatores que poderiam contribuir para o sucesso deste ensaio específico, com base estritamente no contexto fornecido.)

        RESPOSTA ANALÍTICA:
        """
        RAG_PROMPT = PromptTemplate(
            template=prompt_template_rag, input_variables=["context", "question"]
        )

        qa_chain_global = RetrievalQA.from_chain_type(
            llm=llm,
            chain_type="stuff", 
            retriever=retriever,
            return_source_documents=True,
            chain_type_kwargs={"prompt": RAG_PROMPT}
        )
        print("Cadeia RetrievalQA configurada com sucesso.")
    except Exception as e:
        print(f"ERRO ao configurar a cadeia RetrievalQA: {e}")
else:
    if not vector_store_global:
        print("AVISO: Vector store não está disponível. A cadeia RetrievalQA não será configurada.")
    if not api_key:
        print("AVISO: Chave API da OpenAI não está disponível. A cadeia RetrievalQA não será configurada.")
```

```python
app = Flask(__name__)
CORS(app)

@app.route("/", methods=["GET"])
def root_route():
    return "<h3>API RAG para Ensaios Clínicos está a funcionar!</h3>Use o endpoint /v1/predict_trial para interagir."

@app.route("/v1/predict_trial", methods=["POST"])
def handle_predict_trial():
    global qa_chain_global
    try:
        data = request.json
        user_query = data.get("query")

        if not user_query:
            return jsonify({"error": "A 'query' é obrigatória no corpo do pedido."}), 400

        if qa_chain_global is None:
            error_msg = "A cadeia RAG (RetrievalQA) não está inicializada."
            if not vector_store_global:
                error_msg += " O Vector Store não foi carregado/criado."
            if not api_key:
                error_msg += " A chave API da OpenAI não está configurada."
            print(f"Erro no endpoint /v1/predict_trial: {error_msg}")
            return jsonify({"error": error_msg}), 503 # Service Unavailable
        
        print(f"Recebida query para /v1/predict_trial: {user_query}")
        # Usar .invoke() que retorna um dicionário. A resposta está na chave 'result'.
        response_dict = qa_chain_global.invoke({"query": user_query})
        
        final_answer = response_dict.get('result', 'Erro: Nenhuma resposta encontrada na chave \'result\'.')
        source_docs_content = []
        if response_dict.get('source_documents'):
            for doc in response_dict['source_documents']:
                source_docs_content.append({
                    "page_content": doc.page_content,
                    "metadata": doc.metadata
                })

        return jsonify({
            "query": user_query,
            "prediction_explanation": final_answer,
            "source_documents_retrieved": source_docs_content
        })
    
    except Exception as e:
        print(f"Erro ao processar o pedido em /v1/predict_trial: {e}")
        import traceback
        traceback.print_exc() # Imprimir traceback completo no servidor para depuração
        return jsonify({"error": f"Ocorreu um erro interno no servidor: {str(e)}"}), 500

if __name__ == "__main__":
    if not api_key:
        print("### ERRO CRÍTICO: APLICAÇÃO NÃO PODE INICIAR SEM A CHAVE API DA OPENAI ###")
        print("### Por favor, configure API_KEY_PATH corretamente ou defina a variável de ambiente OPENAI_API_KEY ###")
    elif vector_store_global is None:
        print("### ERRO CRÍTICO: APLICAÇÃO NÃO PODE INICIAR SEM UM VECTOR STORE VÁLIDO ###")
        print("### Verifique os logs para erros no carregamento ou criação do índice FAISS. ###")
    elif qa_chain_global is None:
        print("### ERRO CRÍTICO: APLICAÇÃO NÃO PODE INICIAR PORQUE A CADEIA RAG NÃO FOI CONFIGURADA ###")
        print("### Verifique os logs para erros na configuração da RetrievalQA. ###")
    else:
        print("[INFO] A iniciar o servidor Flask para a API RAG de Ensaios Clínicos...")
        # use_reloader=False é importante para evitar que a inicialização do RAG ocorra duas vezes
        app.run(debug=False, port=5000, use_reloader=False) 
```
