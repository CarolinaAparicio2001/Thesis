```python
#from lightning_template.utils.cli import recursive_instantate_class
model_dim = 768

smoe_encoder = SparseMOELayer(
    expert_cfg=FeedForwardLayer,  # pass the class, not a lambda or dict
    num_experts=4,
    input_dim=model_dim,
    topk=2,
)
```

```python
def build_encoder():
    return nn.TransformerEncoder(
        nn.TransformerEncoderLayer(
            d_model=model_dim,
            nhead=8,
            dim_feedforward=2048,
            dropout=0.1,
            activation='relu',
            batch_first=True
        ),
        num_layers=2
    )

input_parts = [
    "table", "summarization", "description", "criteria",
    "smiles", "smiles_concat", "smiles_summarization",
    "drugs", "drugs_concat", "drugs_summarization",
    "diseases", "diseases_concat", "diseases_summarization",
    "smiles_transformer_concat"
]
encoders = nn.ModuleDict({k: build_encoder() for k in input_parts})
```

```python
model = MMCTO(
    encoders=encoders,
    smoe_encoder=smoe_encoder,
    aux_loss=True,
    aux_loss_share_fc=False,
    moe_method="weighted",
    vocab_size=28996,  # BERT-base-cased
    model_dim=model_dim,
)
```

```python
optimizer = torch.optim.Adam(model.parameters(), lr=2e-5)
#scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=5, gamma=0.8)
```

```python
model.train()
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)

for batch in dataloader:
    # Move batch to device if needed
    batch = {k: v.to(device) if torch.is_tensor(v) else v for k, v in batch.items()}
    #model = model.cuda()

    out = model(batch)
    loss = out["loss_dict"]["loss"]

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    print(f"Loss: {loss.item():.4f}")
    break  # remove break to run full epoch
```
