```python
import pandas as pd
import json
```

```python
drug_info_df = pd.read_csv('Data/drugbank_drugs_info.csv')  # Adjust the path as necessario
```

```python
duplicates = drug_info_df[drug_info_df['name'].duplicated(keep=False)]
duplicates
```

```python
# Convert 'name' column to lowercase in the CSV
drug_names_in_csv = set(drug_info_df['name'].dropna().str.lower())
len(drug_names_in_csv)
```

```python
with open('drug_description_hint.json', 'r') as f:
    drug_info = json.load(f)

# Convert drug names in JSON to lowercase during filtering
filtered_drug_info = {drug.lower(): description for drug, description in drug_info.items()
                      if drug.lower() in drug_names_in_csv}

len(filtered_drug_info)
```

```python
missing_in_filtered_info = [drug for drug in drug_names_in_csv if drug.lower() not in filtered_drug_info]

# Print the number of missing drugs
print(f"Drugs from the CSV that are not present in the filtered JSON: {len(missing_in_filtered_info)}")
```

```python
missing_drugs_with_null = {drug.lower(): None for drug in missing_in_filtered_info}

# Combine the filtered JSON data with the missing drugs
combined_drug_info = {**filtered_drug_info, **missing_drugs_with_null}
```

```python
with open('druginfo_hint_final.json', 'w') as f:
    json.dump(combined_drug_info, f, indent=4)

# Print the final count of entries in the combined JSON file
print(f"Combined JSON with {len(combined_drug_info)} drugs saved to 'druginfo_hint_final.json'.")
```
